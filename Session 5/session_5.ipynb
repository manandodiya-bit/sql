# Session 5: WHERE Clause + LIKE (Advanced Filtering & Pattern Matching)

This interactive notebook covers condition-based filtering in SQL: compound Boolean logic (`AND` / `OR`), precedence rules, substring wildcard searches with `LIKE`, range queries with `BETWEEN`, and set membership with `IN`.

---

## Overview of Tasks
- **Task 1:** Create `Restaurants` table (`id`, `name`, `cuisine`, `rating`, `city`), seed 8 realistic Zomato restaurant records across Gujarat.
- **Task 2:** Filter restaurants with `rating > 4.0` in `'Ahmedabad'` or `'Surat'` using compound conditions.
- **Task 3:** Search restaurant names starting with `'Swa'` using the `LIKE 'Swa%'` wildcard.
- **Task 4:** Find restaurants with ratings between `3.5` and `4.5` (inclusive) using `BETWEEN`.
- **Task 5:** Filter specific cuisines (`'Chinese'`, `'Italian'`, `'South Indian'`) using the `IN` operator.

### Task 1: Create 'Restaurants' Table & Seed Records

```python
import sqlite3
import os
import pandas as pd

os.makedirs("database", exist_ok=True)
db_path = os.path.join("database", "restaurants.db")
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

cursor.execute("DROP TABLE IF EXISTS Restaurants;")
create_table_sql = """
CREATE TABLE Restaurants (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name VARCHAR(100) NOT NULL,
    cuisine VARCHAR(50) NOT NULL,
    rating DECIMAL(2, 1) NOT NULL,
    city VARCHAR(50) NOT NULL,
    created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
"""
cursor.execute(create_table_sql)

sample_restaurants = [
    ("Swagat Restaurant", "Gujarati / North Indian", 4.3, "Ahmedabad"),
    ("Swadisht Zaika", "South Indian", 4.1, "Surat"),
    ("Gwalia Sweets", "Chinese", 3.8, "Ahmedabad"),
    ("La Pino'z Pizza", "Italian", 4.4, "Surat"),
    ("Barbeque Nation", "Mughlai / BBQ", 4.6, "Vadodara"),
    ("Sankalp Restaurant", "South Indian", 4.2, "Ahmedabad"),
    ("Swastik Fast Food", "Fast Food", 3.4, "Rajkot"),
    ("Mainland China", "Chinese", 4.5, "Ahmedabad")
]
cursor.executemany(
    "INSERT INTO Restaurants (name, cuisine, rating, city) VALUES (?, ?, ?, ?);",
    sample_restaurants
)
conn.commit()

df_all = pd.read_sql_query("SELECT * FROM Restaurants;", conn)
display(df_all)
```

### Task 2: Filter by Rating (> 4.0) and City ('Ahmedabad' or 'Surat')
> **Important:** Notice the use of parentheses `(city = 'Ahmedabad' OR city = 'Surat')`. Because `AND` has higher precedence than `OR`, omitting parentheses changes the Boolean evaluation logic!

```python
query_t2 = """
SELECT id, name, cuisine, rating, city
FROM Restaurants
WHERE rating > 4.0 AND (city = 'Ahmedabad' OR city = 'Surat')
ORDER BY rating DESC;
"""
df_t2 = pd.read_sql_query(query_t2, conn)
display(df_t2)
```

### Task 3: Pattern Matching with LIKE ('Swa%')
The `%` wildcard matches zero or more characters. `LIKE 'Swa%'` retrieves all restaurants whose names begin with `'Swa'`.

```python
query_t3 = "SELECT id, name, cuisine, rating, city FROM Restaurants WHERE name LIKE 'Swa%';"
df_t3 = pd.read_sql_query(query_t3, conn)
display(df_t3)
```

### Task 4: Range Query with BETWEEN (Rating 3.5 to 4.5)
The `BETWEEN` operator in SQL is **inclusive** on both boundaries (equivalent to `rating >= 3.5 AND rating <= 4.5`).

```python
query_t4 = """
SELECT id, name, cuisine, rating, city
FROM Restaurants
WHERE rating BETWEEN 3.5 AND 4.5
ORDER BY rating DESC;
"""
df_t4 = pd.read_sql_query(query_t4, conn)
display(df_t4)
```

### Task 5: Set Membership Filtering with IN Operator
The `IN` operator concisely tests if a column's value matches any value within a defined list.

```python
query_t5 = """
SELECT id, name, cuisine, rating, city
FROM Restaurants
WHERE cuisine IN ('Chinese', 'Italian', 'South Indian')
ORDER BY cuisine, rating DESC;
"""
df_t5 = pd.read_sql_query(query_t5, conn)
display(df_t5)

conn.close()
```
